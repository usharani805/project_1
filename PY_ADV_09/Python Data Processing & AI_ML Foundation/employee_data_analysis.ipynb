```python
import pandas as pd

# Read employee CSV file
df = pd.read_csv("employees.csv")

print("Employee Data:")
print(df)

print("\nData Information:")
print(df.info())
```

```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Remove duplicate records
df = df.drop_duplicates()

print("\nData after removing duplicates:")
print(df)
```

```python
# Filter IT department employees
it_employees = df[df["Department"] == "IT"]

print("\nIT Department Employees:")
print(it_employees)
```

```python
it_employees = df[df["Department"] == "IT"]

print("\nIT Department Employees:")
print(it_employees)
```

```python
# Filter employees with salary above 50000
high_salary = df[df["Salary"] > 50000]

print("\nEmployees with salary above 50000:")
print(high_salary)
```

```python
# Sort employees by salary in descending order
sorted_employees = df.sort_values(by="Salary", ascending=False)

print("\nEmployees sorted by salary (highest to lowest):")
print(sorted_employees)
```

```python
# Simple Data Cleaning Pipeline

clean_df = df.copy()

# 1. Remove missing values
clean_df = clean_df.dropna()

# 2. Remove duplicate records
clean_df = clean_df.drop_duplicates()

print("\nFinal Clean Dataset:")
print(clean_df)
```

```python
# Save cleaned dataset to CSV
clean_df.to_csv("cleaned_employees.csv", index=False)

print("\nClean CSV file created successfully!")
```

```python
# Basic Statistics for Employee Data

print("\nEmployee Data Statistics:")

print("Total Employees:", len(clean_df))
print("Average Salary:", clean_df["Salary"].mean())
print("Highest Salary:", clean_df["Salary"].max())
print("Lowest Salary:", clean_df["Salary"].min())
print("Total Salary:", clean_df["Salary"].sum())
```

```python
# Create Employee Details DataFrame
employee_details = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Kumar", "Sita", "Arun"]
})

# Create Salary DataFrame
employee_salary = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Department": ["IT", "HR", "Finance", "IT", "HR"],
    "Salary": [50000, 50000, 55000, 60000, 52500]
})

# Merge both DataFrames using ID
merged_data = pd.merge(employee_details, employee_salary, on="ID")

print("Merged Employee Data:")
print(merged_data)
```
